```python
import pandas as pd
import numpy as np

# Import dataset
df = pd.read_csv("RealEstate_Prices.csv")

# Display first 5 rows
print(df.head())
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()
```

```python
# Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df.columns)
```

```python
df["sale_price_inr"]
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
neighborhood_data = pd.DataFrame({
    "neighborhood": [
        "Downtown",
        "Suburb",
        "Riverside",
        "Greenfield",
        "Hillside",
        "Lakeside"
    ],
    "average_income": [
        85000,
        62000,
        70000,
        58000,
        65000,
        76000
    ],
    "schools_nearby": [
        8, 6, 7, 5, 6, 9
    ]
})

print(neighborhood_data)
```

```python
merged_df = pd.merge(
    df,
    neighborhood_data,
    on="neighborhood",
    how="left"
)

print(merged_df.head())
```

```python
df["sale_date"] = pd.to_datetime(df["sale_date"])
```

```python
properties_2023_onwards = df[
    df["sale_date"].dt.year >= 2023
]

print(properties_2023_onwards.head())
```

```python
villas = df[
    df["property_type"] == "Villa"
]

print(villas.head())
```

```python
downtown = df[
    df["neighborhood"] == "Downtown"
]

print(downtown.head())
```

```python
filtered_data = df[
    (df["sale_date"].dt.year >= 2023) &
    (df["property_type"] == "Apartment")
]

print(filtered_data)
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=[
        "city",
        "neighborhood",
        "property_type",
        "condition",
        "parking_available",
        "furnished"
    ],
    drop_first=True
)

print(encoded_df.head())
```

```python
avg_price_neighborhood = (
    df.groupby("neighborhood")["sale_price_inr"]
    .mean()
)

print(avg_price_neighborhood)
```

```python
avg_price_type = (
    df.groupby("property_type")["sale_price_inr"]
    .mean()
)

print(avg_price_type)
```

```python
summary = df.groupby("neighborhood").agg(
    average_price=("sale_price_inr", "mean"),
    median_price=("sale_price_inr", "median"),
    average_area=("area_sq_ft", "mean"),
    property_count=("property_id", "count")
)

print(summary)
```

```python
Q1 = df["sale_price_inr"].quantile(0.25)
Q3 = df["sale_price_inr"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["sale_price_inr"] < lower_limit) |
    (df["sale_price_inr"] > upper_limit)
]

print("Outliers:")
print(outliers)
```

```python
df["sale_price_inr"] = df["sale_price_inr"].clip(
    lower=lower_limit,
    upper=upper_limit
)
```

```python
print("Final dataset shape:")
print(df.shape)

print("\nFinal columns:")
print(df.columns)

print("\nMissing values:")
print(df.isnull().sum())

print("\nFinal dataset:")
print(df.head())
```

```python
df.to_csv(
    "RealEstate_Prices_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
