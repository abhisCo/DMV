```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("City_Air_Quality.csv")

df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Air_Quality"].value_counts()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["City"],
    df["AQI"],
    marker="o"
)

plt.title("AQI Across Different Cities")
plt.xlabel("City")
plt.ylabel("AQI")
plt.xticks(rotation=45)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["City"],
    df["PM2.5"],
    marker="o"
)

plt.title("PM2.5 Levels Across Cities")
plt.xlabel("City")
plt.ylabel("PM2.5")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["City"],
    df["PM10"],
    marker="o"
)

plt.title("PM10 Levels Across Cities")
plt.xlabel("City")
plt.ylabel("PM10")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["City"],
    df["NO2"],
    marker="o"
)

plt.title("NO2 Levels Across Cities")
plt.xlabel("City")
plt.ylabel("NO2")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["City"],
    df["SO2"],
    marker="o"
)

plt.title("SO2 Levels Across Cities")
plt.xlabel("City")
plt.ylabel("SO2")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df["City"],
    df["AQI"]
)

plt.title("AQI Comparison Across Cities")
plt.xlabel("City")
plt.ylabel("AQI")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.boxplot(df["AQI"])

plt.title("AQI Distribution")
plt.ylabel("AQI")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.boxplot(
    [
        df["PM2.5"],
        df["PM10"],
        df["NO2"],
        df["SO2"],
        df["CO"],
        df["O3"]
    ],
    labels=[
        "PM2.5",
        "PM10",
        "NO2",
        "SO2",
        "CO",
        "O3"
    ]
)

plt.title("Pollutant Distribution")
plt.ylabel("Pollutant Level")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["PM2.5"],
    df["AQI"]
)

plt.title("AQI vs PM2.5")
plt.xlabel("PM2.5")
plt.ylabel("AQI")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["PM10"],
    df["AQI"]
)

plt.title("AQI vs PM10")
plt.xlabel("PM10")
plt.ylabel("AQI")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df["City"],
    df["AQI"],
    label="AQI"
)

plt.title(
    "Air Quality Index Across Indian Cities",
    fontsize=16
)

plt.xlabel("City", fontsize=12)
plt.ylabel("AQI", fontsize=12)

plt.xticks(rotation=45)

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["Temperature_C"],
    df["AQI"]
)

plt.title("AQI vs Temperature")
plt.xlabel("Temperature (°C)")
plt.ylabel("AQI")

plt.grid(True)
plt.show()
```
